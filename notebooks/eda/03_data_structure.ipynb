# Cấu trúc dữ liệu gần raw

**Trạng thái: SCAFFOLD — chưa có code thực thi hoặc kết quả.**

Quan sát lớp, giới tính, county và lender; giữ giới hạn mô tả của EDA toàn bộ nguồn.

## Đầu vào và cấu hình

Chưa chọn snapshot/run. Khi triển khai, ghi nguồn, snapshot_id, phiên bản code/config và kernel theo task. Tham số lấy từ configs/; dùng package hmda cho logic dùng chung.

## Thực hiện và kiểm tra

Chưa triển khai. Bổ sung các bước theo thứ tự dữ liệu vào → kiểm tra → phân tích/thực nghiệm → lưu kết quả, trong phạm vi task.

## Kết quả và giới hạn

Chưa có kết quả. Chỉ bổ sung kết luận sau khi chạy và kiểm chứng; liên kết bằng chứng tại agent/RESEARCH_LOG.md.
